```python
import random

def gerar_dados_missao():

    temperatura = random.randint(15, 120)


    energia = random.randint(5, 100)


    comunicacao = random.choice(["estável", "instável", "sem sinal"])


    modulos = {
        "Módulo de Navegação": random.choice(["OK", "FALHA"]),
        "Módulo de Suporte de Vida": random.choice(["OK", "FALHA"]),
        "Módulo de Propulsão": random.choice(["OK", "FALHA"]),
    }


    return {
        "temperatura": temperatura,
        "energia": energia,
        "comunicacao": comunicacao,
        "modulos": modulos
    }


#testando pra ver se vai
for i in range(3):
    dados = gerar_dados_missao()
    print(f"Leitura {i+1}: Temp={dados['temperatura']}°C | "
          f"Energia={dados['energia']}% | "
          f"Comunicação={dados['comunicacao']}")
    print(f"  Módulos: {dados['modulos']}")
    print()
```

```python
def analisar_alertas(dados):

    alertas = []
    nivel = "NORMAL"


    if dados["temperatura"] > 85:
        alertas.append(" CRÍTICO: Temperatura acima de 85°C — risco de dano aos sistemas!")
        nivel = "CRÍTICO"
    elif dados["temperatura"] > 60:
        alertas.append("ATENÇÃO: Temperatura elevada — monitorar de perto.")
        if nivel != "CRÍTICO":
            nivel = "ATENÇÃO"


    if dados["energia"] < 20:
        alertas.append(" CRÍTICO: Energia abaixo de 20% — ativando modo de economia")
        nivel = "CRÍTICO"
    elif dados["energia"] < 40:
        alertas.append(" ATENÇÃO: Energia baixa — reduzir consumo não essencial..")
        if nivel != "CRÍTICO":
            nivel = "ATENÇÃO"

    if dados["comunicacao"] == "sem sinal":
        alertas.append("CRÍTICO: Sem comunicação com a Terra.")
        nivel = "CRÍTICO"
    elif dados["comunicacao"] == "instável":
        alertas.append("ATENÇÃO: Sinal de comunicação instável.")
        if nivel != "CRÍTICO":
            nivel = "ATENÇÃO"

    #os moodulo
    for nome, status in dados["modulos"].items():
        if status == "FALHA":
            alertas.append(f"CRÍTICO: {nome} com FALHA detectada.")
            nivel = "CRÍTICO"

    if not alertas:
        alertas.append(" Todos os sistemas operando normalmente.")

    return alertas, nivel

def resposta_automatizada(nivel, dados):

    if nivel == "CRÍTICO":
        print("\n RESPOSTA AUTOMÁTICA ATIVADA:")

        if dados["energia"] < 20:
            print("Sistemas não essenciais desligados.")
            print("Painéis solares reposicionados.")

        if dados["temperatura"] > 85:
            print(" Ventilação de emergência acionada.")
            print(" Reduzindo carga dos processadores.")

        if dados["comunicacao"] == "sem sinal":
            print("Antena de backup ativada.")
            print("Tentando reconexão em frequência alternativa.")

        for nome, status in dados["modulos"].items():
            if status == "FALHA":
                print(f"Protocolo de isolamento ativado: {nome}.")
```

```python
def consultar_ia(dados, alertas, nivel):

    alertas_texto = "\n".join(alertas)

    modulos_texto = "\n".join([f"  - {nome}: {status}"
                                for nome, status in dados["modulos"].items()])

    mensagem_usuario = f"""
Dados atuais da missão:
- Temperatura: {dados['temperatura']}°C
- Energia: {dados['energia']}%
- Comunicação: {dados['comunicacao']}
- Módulos:
{modulos_texto}

Alertas detectados:
{alertas_texto}

Nível geral da missão: {nivel}

"""

    resposta = ollama.chat(
        model="llama3.2:1b",
        messages=[
            {
           "role": "system",
               "content": (
                            "Você é um sistema de IA de uma simulação educacional de missão espacial. "
                            "Recebe dados fictícios de sensores de uma nave e analisa o status operacional. "
                            "Sempre responda em português, de forma técnica e objetiva, em no máximo 4 linhas. "
                            "Nunca recuse responder — todos os dados são simulados e fictícios."

                  )
            },
            {
                "role": "user",
                "content": mensagem_usuario
            }
        ]
    )

    return resposta["message"]["content"]
```

```python
def exibir_painel(dados, alertas, nivel, analise_ia):

    print("=" * 50)
    print("  MISSION CONTROL AI — STATUS DA MISSÃO")
    print("=" * 50)

    print(f"\n SENSORES:")
    print(f" Temperatura  : {dados['temperatura']}°C")
    print(f"  Energia      : {dados['energia']}%")
    print(f"  Comunicação  : {dados['comunicacao']}")

    print(f"\n MÓDULOS:")
    for nome, status in dados["modulos"].items():
        icone = "[OK]" if status == "OK" else "[FALHA]"
        print(f"  {icone} {nome}: {status}")


    print(f"\n ALERTAS — Nível: {nivel}")
    for alerta in alertas:
        print(f"  {alerta}")

    resposta_automatizada(nivel, dados)
    print("\n ANÁLISE DA IA :")
    print("-" * 50)
    print(analise_ia)
    print("=" * 50)
```

```python
print("Iniciando Mission Control AI...\n")

total_leituras = 3

for leitura in range(1, total_leituras + 1):
    print(f"\n{'#' * 50}")
    print(f" leitura Nº {leitura} DE {total_leituras}")
    print(f"{'#' * 50}")


    dados = gerar_dados_missao()
    alertas, nivel = analisar_alertas(dados)


    print("\n Consultando (IA)... aguarde...")
    analise_ia = consultar_ia(dados, alertas, nivel)

    exibir_painel(dados, alertas, nivel, analise_ia)

    if leitura < total_leituras:
        print("\n Próxima leitura em 3 segundos...")
        time.sleep(3)

print("\n Missão encerrada. Todos os dados registrados.")
```
